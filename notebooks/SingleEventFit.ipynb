# Example: Running a Single Event Fit

This notebook shows a minimal example of how to run the fitEvent Python interface from the Cygno_PMT_fits library

### Import

```python
import fitEvent
```

### Input Data

```python
# PMT collected charges (in nC)
collected_charges = [
    0.01296222395833333,
    0.02641947322970965,
    0.015746124031008,
    0.0081006272672307,
]

# Calibration constants (example)
calibrations = [
    3.205,
    3.545,
    4.512,
    4.138
]
```

### Run the Fit

```python
results = fitEvent.runSingleEvent(collected_charges, calibrations)
```

### Output

```python
print(results)
```

You can also inspec individual fiels: 

```python
results.L_mean, results.x_mean
```
